```python
import pandas as pd
details=pd.read_csv(r'C:\Users\USER\Downloads\ecommerce_order_details.csv')
orders=pd.read_csv(r'C:\Users\USER\Downloads\ecommerce_orders.csv')
```

```python
details.head()
```

```python
import re
details.columns = [re.sub(r'(?<!^)(?=[A-Z])', '_', col).lower() 
              for col in details.columns]
```

```python
details=details.rename(columns={'order _i_d':'order_id','sub-_category':'sub_category'})
details.columns
```

```python
orders.head()
```

```python
import re
orders.columns = [re.sub(r'(?<!^)(?=[A-Z])', '_', col).lower() 
              for col in orders.columns]
```

```python
orders=orders.rename(columns={'order _i_d':'order_id','order _date':'order_date'})
orders.columns
```

```python
orders['Order Date']=pd.to_datetime(orders['Order Date'],dayfirst=True)
orders['month']=orders['Order Date'].dt.month_name()
orders['month_name']=orders['Order Date'].dt.month
```

```python
df=pd.merge(details,orders,on='Order ID',how='inner')
print(df.head())
print(df.info())
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
import re
df.columns = [re.sub(r'(?<!^)(?=[A-Z])', '_', col).lower() 
              for col in df.columns]
```

```python

df=df.rename(columns={'order _i_d':'order_id','sub-_category':'sub_category', 'order _date':'order_date'})
df.columns
```

```python
total_revenue=df['amount'].sum()
total_profit=df['profit'].sum()
total_orders=df['order_id'].nunique()
avg_order_val=total_revenue/total_orders
profit_margin = (total_profit / total_revenue) * 100
print(f"total_revenue      : {total_revenue:.0f}")
print(f"total_profit       : {total_profit:.0f}")
print(f"total_orders       : {total_orders}")
print(f"avg_order_val      :{avg_order_val:.2f}")
print(f" Profit Margin     : {profit_margin:.2f}%")
```

```python
category_summary=df.groupby('category').agg(total_revenue=('amount','sum'),total_profit=('profit','sum'),total_orders=('order_id','count'),avg_quantity=('quantity','mean')).reset_index()
print("\n=== Category Summary ===")
print(category_summary)
```

```python
state_revenue=df.groupby('state')['amount'].sum().sort_values(ascending=False)
print("\n===Top 5 States by Revenue===")
print(state_revenue.head(5))
```

```python
payment_counts=df['payment_mode'].value_counts()
print(payment_counts)
```

```python
loss_orders = df[df['profit'] < 0]
print(f"\n⚠️  Loss-making records: {len(loss_orders)} ({len(loss_orders)/len(df)*100:.1f}%)")
```
